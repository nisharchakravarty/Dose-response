# Formulas
## 0 - Set up initial populations
$ R_0 = \phi * N_0$

$S_0 = (1-\phi)*N_0$

$\phi = $ fraction resistant

## 1 - Set up death rates based on dose response
$ D_r(C) = \frac{Emax_rC}{C + IC50_r}$

$D_s(C) = \frac {Emax_sC}{C + IC50_s}$

## 2 - Derive net growth
$G_{net_r} = G_r - D_r(C)$

$G_{net_s} = G_s - D_s(C)$

## 3- Simulate
$R(t) = R_0e^{G_{net_r}*t}$

$S(t) = S_0e^{G_{net_s}*t}$

## 4 - Compute results (total cell population (concentration of cells/ mL))
$N(t) = R(t) + S(t)$

# Dose Response Formulas
$E(C) = \frac {E_{max}C^\gamma}{EC_{50}^\gamma + C^\gamma}$

$E(C) = \frac {E_{max}C}{EC_{50}+C}$

```python
import math
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib.ticker import FuncFormatter
from scipy.interpolate import RectBivariateSpline
import matplotlib.pyplot as plt

def cellCount(
    time,
    N0,
    phi,
    growth_rate_r,
    growth_rate_s,
    concentration,
    Emax_r,
    Emax_s,
    IC50_r,
    IC50_s,
    hill_slope_r,
    hill_slope_s
):
    def deathRate(
        concentration,
        Emax_r,
        Emax_s,
        IC50_r,
        IC50_s,
    ):
        #(self.Emax_r * C**self.Hill_r) / (C**self.Hill_r + self.IC50_r**self.Hill_r)
        DR_r = (Emax_r*concentration**hill_slope_r)/(concentration**hill_slope_r + IC50_r**hill_slope_r)
        DR_s = (Emax_s*concentration**hill_slope_s)/(concentration**hill_slope_s + IC50_s**hill_slope_s)
        return DR_r, DR_s
        

    DR_r, DR_s = deathRate(concentration,Emax_r,Emax_s, IC50_r,IC50_s)
    #print (DR_r)
    #DR_s = deathRate(concentration, Emax_s, IC50_s)
    gNet_r = growth_rate_r - DR_r
    gNet_s = growth_rate_s - DR_s
   # print(gNet)
    R0 = N0 * phi
    S0 = N0 * (1-phi)
    R_t = R0 * math.exp(gNet_r * time)
    S_t = S0 * math.exp(gNet_s * time)
    N_t = R_t + S_t
    #print(N_t)
    
    return R_t, S_t, N_t

# Constants
t = 72
N0 = 1000
Fr = 0.5 #phi
Gr_r = 0.06
Gr_s = 0.05
C = 0.033
Emaxr = 0
Emaxs = 0.5
ic50r = 10
ic50s = 0.06
hill_r = 1
hill_s = 2

phi_values = np.linspace(0, 0.05, 31)
growth_rate_R_values = np.linspace(0, 0.01, 31)
total_cells_heatmap = np.zeros(
    (len(phi_values),len(growth_rate_R_values))
)
R_results = []
S_results = []
for i, phi_value in enumerate(phi_values):
    for j,growth_rate_R_value in enumerate(growth_rate_R_values):
        R_t, S_t, N_t = cellCount(t, N0, phi_value, growth_rate_R_value, Gr_s, C, Emaxr, Emaxs, ic50r, ic50s, hill_r, hill_s)


        total_cells = N_t
        total_cells_heatmap[i,j] = total_cells

print(total_cells)

df=pd.DataFrame(total_cells_heatmap, index=growth_rate_R_values,columns=phi_values)
# df.head()
ax=sns.heatmap(df)
ax.set_xlabel("Phi")
ax.set_ylabel("Gr values")
ax.set_title("Effect of Gr and phi on Total Cell Count")

ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{phi_values[int(x)]:.3f}")
)

ax.yaxis.set_major_formatter(
    FuncFormatter(lambda y, pos: f"{growth_rate_R_values[int(y)]:.3f}")
)

plt.show()
```

```python
import math
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib.ticker import FuncFormatter
from scipy.interpolate import RectBivariateSpline
import matplotlib.pyplot as plt


#function to calculate total cell count, use parameters to calculate death rate of resistant and sensitive pop
def cellCount(
    time,
    N0,
    phi,
    growth_rate_r,
    growth_rate_s,
    concentration,
    Emax_r,
    Emax_s,
    IC50_r,
    IC50_s,
    hill_slope_r,
    hill_slope_s
):
    def deathRate(
        concentration,
        Emax_r,
        Emax_s,
        IC50_r,
        IC50_s,
    ):
        #(self.Emax_r * C**self.Hill_r) / (C**self.Hill_r + self.IC50_r**self.Hill_r)
        DR_r = (Emax_r*concentration**hill_slope_r)/(concentration**hill_slope_r + IC50_r**hill_slope_r) #calculate death rate resistant
        DR_s = (Emax_s*concentration**hill_slope_s)/(concentration**hill_slope_s + IC50_s**hill_slope_s)
        return DR_r, DR_s
        

    DR_r, DR_s = deathRate(concentration,Emax_r,Emax_s, IC50_r,IC50_s)
    #print (DR_r)
    #DR_s = deathRate(concentration, Emax_s, IC50_s)
    gNet_r = growth_rate_r - DR_r #calculate net growth of resistant based on calculated death rate
    gNet_s = growth_rate_s - DR_s #calculate net growth of sensitive based on calculated death rate
   # print(gNet)
    R0 = N0 * phi
    S0 = N0 * (1-phi)
    R_t = R0 * math.exp(gNet_r * time) #calculate total resistant cell count
    S_t = S0 * math.exp(gNet_s * time) #calculate total sensitive cell count
    N_t = R_t + S_t #total cell count as total of resistant and sensitive cell counts
    #print(N_t)
    
    return R_t, S_t, N_t

# Constants (changeable parameters)
t = 72
N0 = 1000
Fr = 0.5 #phi
Gr_r = 0.06
Gr_s = 0.029
C = 0.033
Emaxr = 0
Emaxs = 0.5
ic50r = 10
ic50s = 0.06
hill_r = 1
hill_s = 2

#y:
phi_values = np.linspace(0, 0.6, 31) #sweep phi values
#x:
growth_rate_R_values = np.linspace(0, 0.01, 31)  #sweep growth rate R values
total_cells_heatmap = np.zeros(
    (len(growth_rate_R_values),len(phi_values))
)
R_results = []
S_results = []
for i, growth_rate_R_value in enumerate(growth_rate_R_values):
    for j,phi_value in enumerate(phi_values):
        R_t, S_t, N_t = cellCount(t, N0, phi_value, growth_rate_R_value, Gr_s, C, Emaxr, Emaxs, ic50r, ic50s, hill_r, hill_s)


        total_cells = N_t
        total_cells_heatmap[j,i] = total_cells

print(total_cells)


        
        

df=pd.DataFrame(total_cells_heatmap, index=phi_values,columns=growth_rate_R_values)

#visualization code

interp = RectBivariateSpline(
        phi_values,
        growth_rate_R_values,
        total_cells_heatmap,
    )

antibody_fine = np.linspace(
    phi_values.min(),
    phi_values.max(),
    200,
)

wait_fine = np.linspace(
    growth_rate_R_values.min(),
    growth_rate_R_values.max(),
    200,
)

Z = interp(
    antibody_fine,
    wait_fine,
)

X, Y = np.meshgrid(wait_fine, antibody_fine)
fig, ax = plt.subplots(
    figsize=(9, 6),
    constrained_layout=True,
)



filled = ax.contourf(
    X,
    Y,
    Z,
    levels=400,
   #vmin = 0, vmax = 20, #rescale colors
    cmap="viridis",
)

contours = ax.contour(
    X,
    Y,
    Z,
    levels=10,
    colors="white",
    linewidths=0.8,
    alpha=0.8,
)

ax.clabel(
    contours,
    inline=True,
    fontsize=8,
    fmt="%.1f",
)


cbar = fig.colorbar(
    filled,
    ax=ax,
)
cbar.set_label(
    "Total Cells",
    fontsize=11,
)

ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x:.3f}")
)

ax.set_xlabel("Growth Rate R (1/hr)")

ax.set_ylabel(
    "Phi",
    fontsize=12,
)

ax.set_title(
    "Total Cells",
    fontsize=14,
    pad=12,
)

plt.show()
```

```python
import numpy as np
import math
import seaborn as sns
import matplotlib.pyplot as plt

def cellCount(
    time,
    N0,
    phi,
    growth_rate_r,
    growth_rate_s,
    concentration,
    Emax_r,
    Emax_s,
    IC50_r,
    IC50_s,
    hill_slope_r,
    hill_slope_s
):
    def deathRate(
        concentration,
        Emax_r,
        Emax_s,
        IC50_r,
        IC50_s,
    ):
        #(self.Emax_r * C**self.Hill_r) / (C**self.Hill_r + self.IC50_r**self.Hill_r)
        DR_r = (Emax_r*concentration**hill_slope_r)/(concentration**hill_slope_r + IC50_r**hill_slope_r)
        DR_s = (Emax_s*concentration**hill_slope_s)/(concentration**hill_slope_s + IC50_s**hill_slope_s)
        return DR_r, DR_s
        

    DR_r, DR_s = deathRate(concentration,Emax_r,Emax_s, IC50_r,IC50_s)
    #print (DR_r)
    #DR_s = deathRate(concentration, Emax_s, IC50_s)
    gNet_r = growth_rate_r - DR_r
    gNet_s = growth_rate_s - DR_s
   # print(gNet)
    R0 = N0 * phi
    S0 = N0 * (1-phi) 
    R_t = R0 * math.exp(gNet_r * time)
    S_t = S0 * math.exp(gNet_s * time)
    N_t = R_t + S_t
    #print(N_t)
    
    return R_t, S_t, N_t

# Constants
t = 24
N0 = 1000
Fr = 0.2 #phi
Gr_r = 0.032
Gr_s = 0.046
Emaxr = 0
Emaxs = 0.251
ic50r = 10
ic50s = 0.33
hill_r = 1
hill_s = 3.2



min_conc= 0
max_conc= 0.06
conc_steps = 30


concentrations = np.linspace(min_conc, max_conc, conc_steps) #sweep concentration
Emin_count = cellCount(t, N0, Fr, Gr_r, Gr_s, 0, Emaxr, Emaxs, ic50r, ic50s, hill_r, hill_s)
Emax_count = cellCount(t, N0, Fr, Gr_r, Gr_s, max_conc, Emaxr, Emaxs, ic50r, ic50s, hill_r, hill_s)
Emin = Emin_count [2]
Emax = Emax_count [2]
print("first Emax is", Emax)




#is cell count at max_conc really the Emax?
conc_test_interval = 0.1
allowable_count_difference = 0.05
conc_unit = (max_conc - min_conc)/conc_steps
almost_max_conc = max_conc *(1-conc_test_interval)
almost_Emax_count= cellCount(t, N0, Fr, Gr_r, Gr_s, almost_max_conc, Emaxr, Emaxs, ic50r, ic50s, hill_r, hill_s)
almost_Emax = almost_Emax_count [2]
cell_count_difference = (almost_Emax - Emax)/Emax
if cell_count_difference > allowable_count_difference:
    Emax = 1000 # change the error handling to something more explicit, this means equilibrium wasn't reached
    print("old emax was", Emax)

E50 = Emax+(Emin-Emax)/2
print("E50 is", E50)

results = []

IC50_set_flag = 0
IC50 = 0

for concentration in concentrations:
    R_t, S_t, N_t = cellCount(t, N0, Fr, Gr_r, Gr_s, concentration, Emaxr, Emaxs, ic50r, ic50s, hill_r, hill_s)
    if N_t< E50 and IC50_set_flag==0:
        IC50_set_flag = 1
        IC50 = concentration

    
    results.append([concentration, R_t, S_t, N_t])

#print values for validation
print("Emin is", Emin)
print("Emax is", Emax)
print("almost Emax is", almost_Emax)
print("cell count difference is", cell_count_difference)
print("allowable count difference is", allowable_count_difference)
print("E50 is", E50)
print("IC50 is", IC50)
print("max conc is", max_conc)
print("almost_max_conc is", almost_max_conc )
    



for concentration, R_t, S_t, N_t in results:
    print(f"{(concentration):8.4f}    {R_t:8.4f}    {S_t:8.4f}    {N_t:8.4f}")

results = np.array(results)

concentration = results[:, 0]
R_t = results[:, 1]
S_t = results[:, 2]
N_t = results[:, 3]


plt.plot(concentration, R_t, label="Resistant", color = "red")
plt.plot(concentration, S_t, label="Sensitive", color = "blue")
plt.plot(concentration, N_t, label="Total", color = "black")

plt.xlabel("Concentration")
plt.ylabel("Cell count")
plt.title("Cells vs Drug Concentration")

plt.xlim(min_conc, max_conc)





plt.legend()
plt.grid(True)
plt.show()


```
